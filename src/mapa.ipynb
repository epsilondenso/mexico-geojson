```python
import folium
import json
```

```python
ruta_test = r"../geojson_nameless/09-Cdmx.geojson"
with open(ruta_test, encoding="utf8") as f:
    geojson = json.load(f)
```

```python
mapa = folium.Map(
    #location=[19.3849, -99.1676],
    zoom_start=11,
    tiles="CartoDB positron"
)
#añadimos el json
folium.GeoJson(
    geojson).add_to(mapa)
```

```python
mapa.save("mapa_test_cdmx.html")
```
